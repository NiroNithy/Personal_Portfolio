```python
import pandas as pd
import numpy as np
from google_play_scraper import app, Sort, reviews_all
import plotly_express as px
```

```python
HNB_project = reviews_all('com.hnb.DigitalApp', sleep_milliseconds=0, lang='en', sort = Sort.NEWEST)
```

```python
df = pd.json_normalize(HNB_project)
```

```python
df.head()
```

```python
columns_to_drop = ["reviewId", "userName", "userImage", "score", "thumbsUpCount", "reviewCreatedVersion", "replyContent","repliedAt"]
df = df.drop(columns=columns_to_drop)
```

```python
df.head()
```

```python
comment_count_by_appversion = df.groupby('appVersion').size().reset_index(name='Total Comments')
```

```python
comment_count_by_appversion = comment_count_by_appversion.sort_values(by='Total Comments', ascending=False)
```

```python
print(comment_count_by_appversion)
```

```python
# Define the list of app versions to filter
desired_versions = ["6.5.2", "3.0.0", "3.7.0", "4.3.2", "4.5.0", "4.6.2", "5.0.0", "5.3.0", "6.2.0", "6.3.0", "6.3.1"]

# Filter the DataFrame to keep only rows with the desired appVersion
filtered_df = df[df['appVersion'].isin(desired_versions)]

# Display the filtered DataFrame
filtered_df.head()
```

```python
from google.colab import sheets
sheet = sheets.InteractiveSheet(df=filtered_df)
```

```python
df['at'] = pd.to_datetime(df['at'])
```

```python
df['year'] = df['at'].dt.year
comment_counts = df.groupby(['appVersion', 'year']).size().reset_index(name='comment_count')
filtered_comments = comment_counts[comment_counts['year'].isin([2024, 2025])]
filtered_comments = filtered_comments.sort_values(by=['year', 'appVersion'], ascending=[False, True])
print(filtered_comments)
```
